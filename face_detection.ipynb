# Face Detection-Based Attendance System

This notebook documents the theoretical foundation and the core components of the Face Detection-Based Attendance System built using OpenCV, Scikit-Learn, and Streamlit.

## 1. Project Overview
The application captures live video from a webcam, detects faces, recognizes them, and automatically marks attendance. It fulfills the following requirements:
- **Live Video Capture:** Using OpenCV `VideoCapture`.
- **Face Detection:** Using Haar Cascade Classifiers.
- **Face Recognition:** Using K-Nearest Neighbors (KNN) algorithm from Scikit-Learn on flattened face image matrices.
- **Attendance Management:** Automatically marking attendance in a CSV file while avoiding duplicate entries within a single session.
- **Streamlit Frontend:** A web interface for face registration, real-time attendance, and report viewing.


## 2. Dataset Creation & Model Training Theory
When a user registers via the Streamlit interface, the webcam captures 100 grayscale face images (resized to 50x50 pixels) and saves them to the `dataset/` directory. 

For training, we read these 50x50 images, flatten them into 1D arrays of size 2500, and train a `KNeighborsClassifier`. KNN is highly effective and lightweight for small to medium-sized face recognition datasets without needing deep learning GPUs.

````python
import cv2
import os
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import joblib

# Function simulating the training process
def train_face_recognition_model(dataset_dir='dataset'):
    faces_data = []
    labels = []
    
    if not os.path.exists(dataset_dir):
        print("Dataset directory not found.")
        return
        
    for student_folder in os.listdir(dataset_dir):
        folder_path = os.path.join(dataset_dir, student_folder)
        if os.path.isdir(folder_path):
            for img_name in os.listdir(folder_path):
                img_path = os.path.join(folder_path, img_name)
                img = cv2.imread(img_path, cv2.IMREAD_GRAYSCALE)
                if img is not None:
                    img_resized = cv2.resize(img, (50, 50))
                    faces_data.append(img_resized.flatten()) # Flatten to 1D
                    labels.append(student_folder)
                    
    if len(faces_data) > 0:
        knn = KNeighborsClassifier(n_neighbors=5)
        knn.fit(faces_data, labels)
        joblib.dump(knn, 'model.pkl')
        print("Model trained and saved successfully.")
    else:
        print("No face data found to train.")

# train_face_recognition_model() # Uncomment to train manually
````
